# Keşifsel Veri Analizi (EDA) ve Özellik Mühendisliği
Bu notebook, birleştirilmiş kullanıcı verisi üzerinde EDA, ilk görselleştirmeler ve feature engineering adımlarını içerir.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Veriyi yükle
df = pd.read_csv('../outputs/merged_data.csv')
df.head()
```

## Olay Türlerinin Dağılımı

```python
plt.figure(figsize=(8,4))
sns.countplot(data=df, x='event_type', order=df['event_type'].value_counts().index)
plt.title('Olay Türlerinin Dağılımı')
plt.xticks(rotation=45)
plt.show()
```

## Kullanıcı Profili Özelliklerinin Dağılımı

```python
profile_cols = ['plan_type', 'country', 'device_type', 'industry']
for col in profile_cols:
    plt.figure(figsize=(6,3))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f'{col} Dağılımı')
    plt.xticks(rotation=45)
    plt.show()
```

## payment_success Olayı Gerçekleştiren Kullanıcıların Analizi

```python
payment_users = df[df['event_type']=='payment_success']['user_id'].unique()
print(f'En az 1 ödeme yapan kullanıcı oranı: {len(payment_users)/df["user_id"].nunique():.2%}')

# Ödeme yapan kullanıcıların profil özellikleri
df_payment = df[df['user_id'].isin(payment_users)]
for col in profile_cols:
    plt.figure(figsize=(6,3))
    sns.countplot(data=df_payment, x=col, order=df_payment[col].value_counts().index)
    plt.title(f'Ödeme Yapanlarda {col} Dağılımı')
    plt.xticks(rotation=45)
    plt.show()
```

## Churn ve Upselling İçin İlk Bulgular

```python
# Son olay tarihine göre kullanıcıların aktiflik analizi
aktiflik = df.groupby('user_id')['timestamp'].max().sort_values()
aktiflik_days = (pd.to_datetime(df['timestamp'].max()) - pd.to_datetime(aktiflik)).dt.days
plt.hist(aktiflik_days, bins=20)
plt.title('Kullanıcıların Son Aktiflik Günleri Dağılımı')
plt.xlabel('Son aktiflikten bu yana geçen gün')
plt.ylabel('Kullanıcı sayısı')
plt.show()
```
